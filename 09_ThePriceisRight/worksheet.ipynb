# 09 The Price is Right

## Your Objective

Your dataset contains three tables from a pizza restaurant:

1. A transactions table with order date, product, and quantity
2. A products table with the current price for each product
3. A price history table with the price changes for each product over time

Your task is to look up the price for each product in the transactions table at that point in time, accurately reflecting the price changes.

NOTE: If the effective date of a price change is equal to the order date for a transaction, then that is the price that should be reflected.

See example below:

![image](../images/ShnCDNAcigFpLy8IKWGLTufQo0.avif)

```python
import pandas as pd

price_history = pd.read_csv('data/price_history.csv', parse_dates=['effective_date'])
price_history.head()
```

```python
transactions = pd.read_csv('data/transactions.csv', parse_dates=['order_date'])
transactions.head()
```

```python
transactions_total = pd.merge_asof(
    transactions, 
    price_history, 
    left_on='order_date', 
    right_on='effective_date',
    by='pizza_id'
).assign(line_total = lambda x: x.price * x.quantity).groupby('order_id').agg({'line_total': 'sum'}).reset_index().rename(columns={'line_total': 'transaction_total'})

transactions_total.head()
```

```python
transactions_total["transaction_total"].sum()
```
