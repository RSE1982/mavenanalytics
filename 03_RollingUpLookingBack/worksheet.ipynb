# 03 Rolling Up, Looking Back

## Your Objective

You’ve been given a table with raw data for individual coffee shop transactions. Your task is to aggregate the total sales by month and store, then calculate the month-over-month change in sales for each store (in dollars).

Example input and output:

![image](../images/KzJBIHeseB8isWzTTTEaMkobA.avif)

```python
import pandas as pd

df = pd.read_csv('data/CoffeeShopSales.csv', parse_dates=['date'])
df.head()
```

```python
df.groupby(['store', df['date'].dt.month]).agg({'sales': 'sum'}).reset_index().rename(columns={'date': 'month'})
```

```python
may_astoria = df[(df['store'] == 'Astoria') & (df['date'].dt.month == 5)]
april_astoria = df[(df['store'] == 'Astoria') & (df['date'].dt.month == 4)]

difference = may_astoria['sales'].sum() - april_astoria['sales'].sum()
print(f"The difference in sales for Astoria between May and April is: ${difference:.2f}")
```
