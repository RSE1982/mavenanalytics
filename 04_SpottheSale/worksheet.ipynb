# 04 Spot the Sale

## Your Objective

You've been given two tables:

1. A table with promotional periods, each with a start and end date
2. A table with orders, each with an order date and quantity

Your task is to join each transaction to the promotion active on its order date.

![image](../images/UzxnkQNwlGdRaDL3M993Bv2Zo.avif)

```python
import pandas as pd

promotions = pd.read_csv('data/promotions.csv', parse_dates=['start_date', 'end_date'])
orders = pd.read_csv('data/orders.csv', parse_dates=['order_date'])
```

```python
df = pd.merge_asof(
    orders.sort_values('order_date'),
    promotions.sort_values('start_date'),
    left_on='order_date',
    right_on='start_date',
    direction='backward'
).query('order_date <= end_date')
df.head()
```
